This file will take in the population df for the hole at each epsilon value (returned by array_job_pop.sh) and concatonate them into a df

```python
#import packages
import geopandas as gpd
import pandas as pd
```

```python
# input the df with hole Multipolygon information for each epsilon value
file_name = input()
```

```python
#convert the inputted file into a path that will allow you to access the population matrices
file_path_as_list = file_name.split('/')[:-1]
file_title = file_name.split('/')[-1]
file_title = file_title.split('.')[0]
file_path = '/'.join(file_path_as_list)
file_path = f"{file_path}/{file_title}"
print(file_path)
```

```python

data = gpd.read_file(file_name)

#create empty df to which we can save each population matrix 
output = pd.DataFrame(data = [], columns = ['eps', 'pop', 'area'])
#for every epsilon value, import the associated population file and concatonate it to the empty data frame produced above
for i in range(len(data)):
    concat_df = pd.read_csv(f"{file_path}/pop_res{i}.csv")
    output = pd.concat([output, concat_df], ignore_index=True)
    print(f"Done with node: {i}")

#save
output.to_csv(f"{file_path}/pop_full.csv")
```
