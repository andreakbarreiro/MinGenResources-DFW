```python
file_path_as_list = file_name.split('/')[:-1]
file_title = file_name.split('/')[-1]
file_title = file_title.split('.')[0]
file_path = '/'.join(file_path_as_list)
file_path = f"{file_path}/{file_title}"
```

```python
print(file_path)
```

```python
def import_and_concat(mat_type, dist_type = 'dists'):
    #loading in the 0th row to get the matrix length 
    test = np.load(f"{file_path}/{mat_type}_cols/{mat_type}_col_0.npz")[f"{dist_type}_{mat_type}"]
    test_shape = test.shape
    
    full_array = np.zeros((test_shape[0], test_shape[0]))
    print('full array size: ', full_array.shape)
    for i in range(test_shape[0]):
        loaded_np = np.load(f"{file_path}/{mat_type}_cols/{mat_type}_col_{i}.npz")
        col = loaded_np[f"{dist_type}_{mat_type}"]
        col = col.reshape((test_shape[0],))
        # print(dist_col)
        # print(time_col) 
        full_array[i:] = col

    # file_path_walk = Path(f"../out_data_{job_id}/walk_cols")
    # file_path_walk.mkdir(parents=True, exist_ok=True)
    
    # np.savez(f"{file_path_walk}/walk_col_{task_id}", walk_speed = walk_speed, dists_walk = dists_walk, time_walk = time_walk)
    
    return full_array
```

```python
def run_all_concats():
    #file_path = f"{file_path}"
    #file_path_walk.mkdir(parents=True, exist_ok=True)
    
    data = {'dists_walk': None, 'time_walk': None, 
           'dists_geod': None, 'time_geod': None, 
           'dists_drive': None, 'time_drive':None}
    
    for i in ['geod']:
    #, 'walk', 'drive']:
        for j in ['dists', 'time']:
            data[f'{j}_{i}'] = import_and_concat(i, j)
        print(i, "done")
    return data
```
