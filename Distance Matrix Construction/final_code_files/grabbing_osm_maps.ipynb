```python
#import packages
import osmnx as ox
import osmnx.routing as routing
import pickle 
```

```python

def load_map(network_type, 
             city = "Dallas County",
             use_bbox = True,
             bbox = [-97.1, 32.5, -96.45, 33.1]):

    """
    Load the osm street map for driving or walking distance and isochrone calculations

    Args:
    network_type (str): 'walk' or 'drive'
    city (str): the U.S. city of interest
    use_bbox (Bool): whether or not to use the bounding box defined (we use a bbox because Dallas City spans two counties
        and where certain points map to end up wonky if we don't account for this)
    bbox (list): the bounding box of interest
    
    """
    # use default bbox to include the southern piece of collin county, if desired
    if use_bbox:
        G = ox.graph_from_bbox(bbox, network_type = network_type, simplify = False)
    else: 
        place = city
        G = ox.graph_from_place(place, network_type = network_type, simplify = False)
    return G
```

```python
G_walk = load_map(network_type = "walk")
print('loaded walk')
G_drive = load_map(network_type = "drive")
print('loaded drive')
```

```python
#pickle the files in order to save them 
with open("Dallas_bbox_walk.pkl", "wb") as file_walk:
    pickle.dump(G_walk, file_walk)
with open("Dallas_bbox_drive.pkl", "wb") as file_drive:
    pickle.dump(G_drive, file_drive)
```
