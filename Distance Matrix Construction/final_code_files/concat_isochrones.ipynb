This script takes the individual isochrone matrices returned by array_job_iso.sh/iso_slurm.py and concatonates them into a single data frame where each row is a single resource location/point in the original resource data frame

```python
#import necessary packages
import geopandas as gpd
import pandas as pd
```

```python
# input the original data file with resource locations
file_name = input()

# type_ can be walk or drive depending on whether you are interested in producing a walk or drive isochrone data frame
type_ = input()
```

```python
#convert the inputted file into a path that will allow you to access the isochrone matrices
file_path_as_list = file_name.split('/')[:-1]
file_title = file_name.split('/')[-1]
file_title = file_title.split('.')[0]
file_path = '/'.join(file_path_as_list)
file_path = f"{file_path}/{file_title}/isochrones_{type_}"
#test that the file path looks correct
print(file_path)
```

```python
#load in the original data file with resource locations in order to determine how many resources (i.e. isochrone matrices) that you're working with
data = gpd.read_file(file_name)

#create empty gdf to which we can save each isochrone matrix 
output = gpd.GeoDataFrame(data = [], columns = ['node', 'eps', 'geometry'])

#for every resource in the orginal data file, load the associated isochrone matric as a gdf, then concat that gdf with the empty one created above
for i in range(len(data)):
    concat_df = gpd.read_file(f"{file_path}/node{i}.geojson")
    output = pd.concat([output, concat_df], ignore_index=True)
    print(f"Done with node: {i}")

#save the file 
output.to_file(f"{file_path}/isochrone_full.geojson", driver = 'GEOJSON')
```
