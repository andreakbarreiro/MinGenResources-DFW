```python
# Read in the grid intersected polygons obtained from the ArcGIS process
grid_intersect = gpd.read_file("data/grid_intersection/grid_intersect.geojson")
# change to crs used for Open Street Map API
grid_intersect = grid_intersect.to_crs("EPSG:4326")
# calculate centroids
grid_intersect["centroid"] = grid_intersect.centroid
```

```python
# Just select the columns we want
grid_intersect_centroids = grid_intersect[["NAME", "OWNER", "NOTES", "CITY", "FEATTYPE", "centroid"]]
# write to new file with the centroids
#grid_intersect_centroids.to_file("grid_intersect_centroids.shp")
```

```python
# For the purposes of mapping, change the the CRS used for mapping
grid_intersect_map = grid_intersect.to_crs("EPSG:3857")
# recalculate centroids
grid_intersect_map["centroid"] = grid_intersect_map.centroid
```

```python
# Visualizing how the new centroids look with the grid intersection for the Trinity River Greenbelt
ax = grid_intersect_map[grid_intersect_map["OBJECTID"] == 1919].plot(color="darkolivegreen", edgecolor="black")
ax = grid_intersect_map[grid_intersect_map["OBJECTID"] == 1919].set_geometry("centroid").plot(ax=ax, color = "red", markersize=15)
cx.add_basemap(ax)
ax.set_title("Trinity River Greenbelt")
ax.set_axis_off();
```
