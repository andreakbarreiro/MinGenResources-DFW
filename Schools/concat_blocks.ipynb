This file concatonates the matrix returned by block_labeler.py for each block and returns a full matrix 

```python
import pandas as pd
```

```python
#read in the zeroeth epsilon values' information 
df = pd.read_csv('Schools/block_by_eps/block_label_0.csv')
df.drop('Unnamed: 0', axis = 1, inplace = True)
df = df.set_index('GEOID20')
```

```python
#for every epsilon value, grab the assocaiated block_label csv file and append it to the zeroeth epsilon df
for i in range(1, 163): 
    new_df = pd.read_csv(f'Schools/block_by_eps/block_label_{i}.csv')
    new_df.drop('Unnamed: 0', axis = 1, inplace = True)
    new_df = new_df.set_index('GEOID20')
    df = pd.concat([df, new_df], axis = 1)
   # df = pd.merge(df, new_df, left_on = 'GEOID20', right_on = 'GEOID20', how = 'full')
```

```python
#save
df.to_csv('Schools/block_by_eps/blocks_by_eps_full.csv')
```
